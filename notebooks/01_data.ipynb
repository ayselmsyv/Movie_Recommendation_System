```python
movies["year"] = movies["title"].str.extract(r"\((\d{4})\)")
movies["year"] = movies["year"].astype("Int64")
movies.sort_values("year", ascending=False)[
    ["movieId", "title", "year"]
].head(10)
```

```python
# updating values of Movies if Year < 2015
movies["year"] = movies["title"].str.extract(r"\((\d{4})\)")
movies["year"] = movies["year"].astype("Int64")
movies.loc[(movies.year >= 2015), "Movies"] = "New Movie"
display(movies.tail(10))
```

```python
for index,row in movies.iterrows():
    print(f"Updated row {index} to New Movie")
    print(row)
    break
```
